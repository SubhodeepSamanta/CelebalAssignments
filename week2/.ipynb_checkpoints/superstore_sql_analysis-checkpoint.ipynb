# Superstore Sales Analysis (SQL)

Dataset: Sample Superstore (Kaggle). Using MySQL through Python to filter, aggregate and pull out some business level insights from the order data.

```python
import pandas as pd
import pymysql
from sqlalchemy import create_engine, text
```

## Connecting to MySQL and loading the data

Change the password below to your own MySQL root password (or whichever user you normally connect with). If your MySQL user has no password, leave it as an empty string.

```python
mysql_user = "root"
mysql_password = "root"
mysql_host = "localhost"
mysql_port = 3306
db_name = "superstore_db"

engine_no_db = create_engine(f"mysql+pymysql://{mysql_user}:{mysql_password}@{mysql_host}:{mysql_port}/")
with engine_no_db.connect() as conn:
    conn.execute(text(f"CREATE DATABASE IF NOT EXISTS {db_name}"))
    conn.commit()

engine = create_engine(f"mysql+pymysql://{mysql_user}:{mysql_password}@{mysql_host}:{mysql_port}/{db_name}")
```

```python
df = pd.read_csv("superstore.csv", encoding="latin1")
df.columns = [c.strip().replace(" ", "_").replace("-", "_") for c in df.columns]
df["Order_Date"] = pd.to_datetime(df["Order_Date"], format="%m/%d/%Y")
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"], format="%m/%d/%Y")
df.head()
```

```python
df.to_sql("orders", engine, if_exists="replace", index=False)
print(len(df), "rows loaded into the orders table")
```

## Checking the table

Quick look at the schema and a few sample rows to make sure everything loaded the way it should.

```python
schema = pd.read_sql("DESCRIBE orders", engine)
schema
```

```python
pd.read_sql("SELECT * FROM orders LIMIT 5", engine)
```

```python
pd.read_sql("SELECT COUNT(*) AS total_rows FROM orders", engine)
```

## Filtering with WHERE

Some basic filters first - by region, category, a date range, and a sales threshold.

```python
q = """
SELECT Order_ID, Customer_Name, Sales, Region
FROM orders
WHERE Region = 'West' AND Sales > 1000
ORDER BY Sales DESC
LIMIT 10
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT COUNT(*) AS tech_orders_2016, SUM(Sales) AS tech_sales_2016
FROM orders
WHERE Category = 'Technology' AND Order_Date BETWEEN '2016-01-01' AND '2016-12-31'
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT Order_ID, Customer_Name, Category, Sub_Category, Discount, Profit
FROM orders
WHERE Discount > 0.3 AND Profit < 0
ORDER BY Discount DESC
LIMIT 10
"""
pd.read_sql(q, engine)
```

## Aggregations with GROUP BY

Total sales, quantity and average order value broken down by category, and then by region + category together.

```python
q = """
SELECT Category,
       ROUND(SUM(Sales), 2) AS total_sales,
       SUM(Quantity) AS total_quantity,
       ROUND(AVG(Sales), 2) AS avg_sale,
       COUNT(*) AS num_orders
FROM orders
GROUP BY Category
ORDER BY total_sales DESC
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT Region, Category, ROUND(SUM(Sales), 2) AS total_sales
FROM orders
GROUP BY Region, Category
ORDER BY Region, total_sales DESC
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT Segment, ROUND(SUM(Sales), 2) AS total_sales, ROUND(SUM(Profit), 2) AS total_profit
FROM orders
GROUP BY Segment
ORDER BY total_sales DESC
"""
pd.read_sql(q, engine)
```

## Sorting and limiting - top products, categories etc

```python
q = """
SELECT Product_Name, ROUND(SUM(Sales), 2) AS total_sales
FROM orders
GROUP BY Product_Name
ORDER BY total_sales DESC
LIMIT 10
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT Sub_Category, ROUND(SUM(Profit), 2) AS total_profit
FROM orders
GROUP BY Sub_Category
ORDER BY total_profit DESC
LIMIT 10
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT Ship_Mode, COUNT(*) AS num_orders, ROUND(AVG(Sales), 2) AS avg_sale
FROM orders
GROUP BY Ship_Mode
ORDER BY num_orders DESC
"""
pd.read_sql(q, engine)
```

## Business use cases

### Monthly sales trend

```python
q = """
SELECT YEAR(Order_Date) AS yr, MONTH(Order_Date) AS mo, ROUND(SUM(Sales), 2) AS total_sales
FROM orders
GROUP BY yr, mo
ORDER BY yr, mo
"""
monthly = pd.read_sql(q, engine)
monthly.head(15)
```

```python
monthly.sort_values("total_sales", ascending=False).head(5)
```

### Top customers by total spend

```python
q = """
SELECT Customer_Name, ROUND(SUM(Sales), 2) AS total_spent, COUNT(DISTINCT Order_ID) AS num_orders
FROM orders
GROUP BY Customer_Name
ORDER BY total_spent DESC
LIMIT 10
"""
pd.read_sql(q, engine)
```

### Checking for duplicate order lines

Same Order_ID + Product_ID showing up more than once - could be genuine repeat line items or a data entry issue.

```python
q = """
SELECT Order_ID, Product_ID, COUNT(*) AS times_repeated
FROM orders
GROUP BY Order_ID, Product_ID
HAVING COUNT(*) > 1
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT Order_ID, Product_ID, Sales, Quantity, Discount, COUNT(*) AS c
FROM orders
GROUP BY Order_ID, Product_ID, Sales, Quantity, Discount
HAVING c > 1
"""
pd.read_sql(q, engine)
```

## Validating the results

Row counts, null checks, and a basic sanity check on profit/discount ranges.

```python
pd.read_sql("SELECT COUNT(*) AS row_count FROM orders", engine)
```

```python
q = """
SELECT
  SUM(CASE WHEN Sales IS NULL THEN 1 ELSE 0 END) AS null_sales,
  SUM(CASE WHEN Customer_Name IS NULL THEN 1 ELSE 0 END) AS null_customer,
  SUM(CASE WHEN Order_Date IS NULL THEN 1 ELSE 0 END) AS null_date
FROM orders
"""
pd.read_sql(q, engine)
```

```python
q = """
SELECT MIN(Discount) AS min_discount, MAX(Discount) AS max_discount,
       MIN(Profit) AS min_profit, MAX(Profit) AS max_profit,
       COUNT(CASE WHEN Profit < 0 THEN 1 END) AS loss_making_orders
FROM orders
"""
pd.read_sql(q, engine)
```

## Insights

- Technology brings in the highest total sales of the three categories, but Office Supplies has by far the most orders and units sold - lots of small, frequent purchases.
- The West and East regions are the strongest performers overall, Central is the weakest.
- November and December tend to have the highest monthly sales (year end push), 2017 was the strongest year in the dataset.
- A small number of customers (like Sean Miller and Tamara Chand) account for a disproportionately large share of total spend.
- Copiers and Phones are the most profitable sub-categories by far, even though they don't have the highest unit volume.
- 1871 of the 9994 orders are loss making, almost all of them tied to high discount rates above 30%, which is worth flagging since heavy discounting is eating into margins on those lines.
- No nulls anywhere in the dataset, no exact duplicate order/product/amount combinations, only a handful of repeated Order_ID + Product_ID pairs which look like legitimate separate line entries rather than data errors.
